```python
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
```

```python
caminho = '../data/raw/open_meteo_cuiaba.csv'
df = pd.read_csv(caminho)
df.head()
```

```python
df.info()
```

```python
df.describe(include='all')
```

```python
df.isna().sum()
```

```python
df['Data'] = pd.to_datetime(df['Data'])

plt.figure(figsize=(10, 5))
plt.plot(df['Data'], df['Temperatura Máxima (°C)'], marker='o', label='Máxima')
plt.plot(df['Data'], df['Temperatura Mínima (°C)'], marker='o', label='Mínima')
plt.xlabel('Data')
plt.ylabel('Temperatura (°C)')
plt.title('Temperaturas previstas para Cuiabá/MT')
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
